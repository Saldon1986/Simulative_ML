Постройте выборку бета-распределения объемом 2000 с параметрами a=2, b=6, смещением 3 и разбросом 10. 

Нарисуйте гистограмму для выборки и теоретическую плотность вероятности распределения на одной оси.

Какой график ближе к полученную результату?

```python
a, b, loc, scale, size = 2, 6, 3, 10, 2000

data = beta.rvs(a=a, b=b, loc=loc, scale=scale, size=size)

x = np.linspace(loc, loc + scale, 1000)

pdf = beta.pdf(x, a, b, loc=loc, scale=scale)

plt.hist(data, bins=50, density=True, label="Choise", alpha=0.5)
plt.plot(x, pdf, label="Theoretical probability")
plt.legend()
plt.show();
```

Определите 1, 55 и 90 процентили этого теоретического бета-распределения.

Округлите значения до сотых и запишите в строку через пробел с разделителем разрядов - точкой.

```python
" ".join(
    [f"{x:.2f}" for x in beta.ppf([0.01, 0.55, 0.9], a=a, b=b, loc=loc, scale=scale)]
)
```

Найдите теоретические характеристики распределения.

Результат округлите до сотых выпишете через пробел по порядку: среднее медиана дисперсия стандартное отклонение.

```python
avg = beta.mean(a=a, b=b, loc=loc, scale=scale)
median = beta.median(a=a, b=b, loc=loc, scale=scale)
var = beta.var(a=a, b=b, loc=loc, scale=scale)
std = beta.std(a=a, b=b, loc=loc, scale=scale)

print(" ".join([f"{x:.2f}" for x in (avg, median, var, std)]))
```
