Сформируйте кластеризованную выборку из 20 пользователей из таблицы USERS так, чтобы в нее входили клиенты из 3 случайных компаний (company_id) в таких же пропорциях, как они представлены в генеральной совокупности.

```python
import pandas as pd
import numpy as np
from datetime import datetime, date
import random
```

```python
USERS = pd.read_csv('/content/itresume-users.csv', parse_dates=['date_joined'], date_format='%Y-%m-%d %H:%M:%S')
```

```python
USERS.head(3)
```

```python
# проверим, что есть 20 записей по случайны компаниям
tot = 0
N = 20
while tot < N:
  # series of random companies
  comps = random.sample(USERS['company_id'].unique().tolist(), 3)

  # sampling
  DF = USERS[USERS['company_id'].isin(comps)]
  tot = DF['id'].count()
```

```python
all_columns = USERS.columns.tolist()
users_sample = DF.groupby('company_id', group_keys=False)[all_columns].apply(lambda x: x.sample(int(np.rint(N*len(x)/len(DF)))), include_groups=False)
```

```python
users_sample
```

Далее на основании этой кластерной выборки клиентов сформируйте стратифицированную выборку для таблицы CODERUN таким образом, чтобы в нее входило по 1 случайной строке для каждой даты (created_at).

```python
CODE_RUN = pd.read_csv('/content/itresume-coderun.csv') #, parse_dates=['created_at'], date_format='%Y-%m-%d %H:%M:%S.%f')
CODE_RUN['created_at'] = pd.to_datetime(CODE_RUN['created_at'], format='%Y-%m-%d %H:%M:%S.%f').dt.date
```

```python
CODE_RUN.head(3)
```

```python
CODE_RUN.columns
```

```python
# concat DFs

df_tot = users_sample.merge(CODE_RUN, how='inner', right_on='user_id', left_on='id')
df_tot.columns = ['user_id', 'username', 'date_joined', 'company_id', 'problem_id', 'created_at',
        'problem_id', 'user_id_todrop', 'language_id']
df_tot.drop('user_id_todrop', axis=1)
```

```python
all_cols = df_tot.columns.to_list()
coderun_sample = df_tot.groupby('created_at', group_keys=False)[all_cols].apply(lambda x: x.sample(1))
```
